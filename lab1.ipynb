```python
#Q2
def isPrime(n):
    correct=False
    for i in range(2,n):
        if(n%i==0):
            correct=True
            break
    if(correct):
        return False
    else:
        return True
isPrime(14)
```

```python
# Q22
Students = pd.DataFrame({
    "sid": [1, 2, 3],
    "name": ["Deepak", "Najaf", "Sandy"]
})

Attendance = pd.DataFrame({
    "sid": [1, 2, 3],
    "Attendance": [70, 87, 89]
})

final = pd.merge(Students, Attendance, on="sid")
print(final)
result = final.loc[final["Attendance"] < 75, ["name", "Attendance"]]
print(" ")
print(result)
```

```python
#Section D — Matplotlib (Q23–Q25)
#Q23
import matplotlib.pyplot as plt
avg_chart=df.groupby("section")["Marks"].mean()
plt.bar(avg_chart.index,avg_chart.values)
plt.xlabel("Sections")
plt.ylabel("Marks Average")
plt.title("Marks average per sections")
plt.show()
```
